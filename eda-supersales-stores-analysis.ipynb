## 5.6.1 Outlier Detection using IQR

The Interquartile Range (IQR) method identifies outliers based on the spread of the middle 50% of the data.

- Q1 = 25th percentile
- Q3 = 75th percentile
- IQR = Q3 - Q1
- Lower Bound = Q1 - 1.5 × IQR
- Upper Bound = Q3 + 1.5 × IQR

Any value below the lower bound or above the upper bound is considered an outlier.

```python
# Numerical columns for outlier detection
numerical_features = ["Sales", "Quantity", "Discount", "Profit"]

# Detect outliers using IQR
iqr_outlier_summary = []

for column in numerical_features:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]
    
    iqr_outlier_summary.append({
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers)
    })

iqr_outlier_summary = pd.DataFrame(iqr_outlier_summary)

iqr_outlier_summary
```

## 5.6.2 Outlier Detection using Z-Score

The Z-score indicates how far a value is from the mean in terms of standard deviations.

A common rule is:

- Z-score > +3 → Potential outlier
- Z-score < -3 → Potential outlier
- -3 ≤ Z-score ≤ +3 → Generally considered normal

The Z-score method is applied to Sales, Quantity, Discount, and Profit.

```python
import numpy as np

# Detect outliers using Z-score
zscore_outlier_summary = []

for column in numerical_features:
    mean = df[column].mean()
    std = df[column].std()
    
    z_scores = (df[column] - mean) / std
    
    outliers = df[abs(z_scores) > 3]
    
    zscore_outlier_summary.append({
        "Column": column,
        "Mean": mean,
        "Standard Deviation": std,
        "Outlier Count": len(outliers)
    })

zscore_outlier_summary = pd.DataFrame(zscore_outlier_summary)

zscore_outlier_summary
```

## 5.6.3 Outlier Summary

The results obtained from the IQR and Z-score methods are compared to understand the number of potential outliers in each numerical attribute.

```python
# Create a combined outlier summary

outlier_summary = pd.DataFrame({
    "Column": numerical_features,
    "IQR Outliers": iqr_outlier_summary["Outlier Count"].values,
    "Z-Score Outliers": zscore_outlier_summary["Outlier Count"].values
})

outlier_summary
```
