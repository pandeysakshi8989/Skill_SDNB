```python
# ============================================================
# WEEK 5: HEALTHCARE DATA ANALYTICS
# Data Understanding, Cleaning & Feature Engineering
# ============================================================

# 1. Import Required Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# 2. Loading the Dataset
# ============================================================

# HEALTHCARE DATASET

import kagglehub

# Download latest version
path = kagglehub.dataset_download("prasad22/healthcare-dataset")

print("Path to dataset files:", path)

import os

print(os.listdir(path))

import pandas as pd

df = pd.read_csv(os.path.join(path, "healthcare_dataset.csv"))

print( "Dataset loaded successfully")

# import os
# os.getcwd()

# df = pd.read_csv(r"/kaggle/input/datasets/prasad22/healthcare_dataset.csv")

print("FIRST 5 RECORDS")
display(df.head())

print("\nLAST 5 RECORDS")
display(df.tail())

print("\nRANDOM 5 RECORDS")
display(df.sample(5))


# ============================================================
# 3. Understanding the Dataset
# ============================================================

print("\nDATASET SHAPE:")
print(df.shape)

print("\nNUMBER OF ROWS:", df.shape[0])
print("NUMBER OF COLUMNS:", df.shape[1])

print("\nCOLUMN NAMES:")
print(df.columns.tolist())

print("\nDATA TYPES:")
print(df.dtypes)

print("\nDATASET INFORMATION:")
df.info()

print("\nSTATISTICAL SUMMARY:")
display(df.describe(include="all"))


# ============================================================
# 4. Identifying Data Types
# ============================================================

numerical_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

categorical_columns = df.select_dtypes(
    include="object"
).columns.tolist()

print("\nNUMERICAL ATTRIBUTES:")
print(numerical_columns)

print("\nCATEGORICAL ATTRIBUTES:")
print(categorical_columns)


# ============================================================
# 5. Missing Value Analysis
# ============================================================

print("\nMISSING VALUES:")
display(df.isnull().sum())

missing_percentage = (
    df.isnull().sum() / len(df)
) * 100

missing_summary = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": missing_percentage
})

print("\nMISSING VALUE SUMMARY:")
display(missing_summary)

# Remove rows containing missing values
df = df.dropna()

print("\nMISSING VALUES AFTER CLEANING:")
display(df.isnull().sum())


# ============================================================
# 6. Duplicate Analysis
# ============================================================

print("\nNUMBER OF DUPLICATE RECORDS:")
print(df.duplicated().sum())

# Remove duplicate records
df = df.drop_duplicates()

print("\nDUPLICATES AFTER REMOVAL:")
print(df.duplicated().sum())


# ============================================================
# 7. Date Preprocessing
# ============================================================

df["Date of Admission"] = pd.to_datetime(
    df["Date of Admission"],
    errors="coerce"
)

df["Discharge Date"] = pd.to_datetime(
    df["Discharge Date"],
    errors="coerce"
)

print("\nDATE COLUMN DATA TYPES:")
print(df[[
    "Date of Admission",
    "Discharge Date"
]].dtypes)

# Extract useful date components
df["Admission Year"] = df["Date of Admission"].dt.year
df["Admission Month"] = df["Date of Admission"].dt.month

print("\nDATE ATTRIBUTES:")
display(df[[
    "Date of Admission",
    "Discharge Date",
    "Admission Year",
    "Admission Month"
]].head())


# ============================================================
# 8. Numeric Data Conversion
# ============================================================

df["Age"] = pd.to_numeric(
    df["Age"],
    errors="coerce"
)

df["Billing Amount"] = pd.to_numeric(
    df["Billing Amount"],
    errors="coerce"
)

df["Room Number"] = pd.to_numeric(
    df["Room Number"],
    errors="coerce"
)


# ============================================================
# 9. Categorical Data Cleaning
# ============================================================

# Remove unwanted spaces
text_columns = [
    "Gender",
    "Medical Condition",
    "Admission Type",
    "Insurance Provider",
    "Blood Type",
    "Medication",
    "Test Results"
]

for column in text_columns:
    df[column] = df[column].str.strip()

# Standardize text formatting
df["Gender"] = df["Gender"].str.title()
df["Medical Condition"] = df["Medical Condition"].str.title()
df["Admission Type"] = df["Admission Type"].str.title()
df["Insurance Provider"] = df["Insurance Provider"].str.title()
df["Blood Type"] = df["Blood Type"].str.upper()
df["Medication"] = df["Medication"].str.title()
df["Test Results"] = df["Test Results"].str.title()

print("\nUNIQUE GENDER VALUES:")
print(df["Gender"].unique())

print("\nUNIQUE MEDICAL CONDITIONS:")
print(df["Medical Condition"].unique())

print("\nUNIQUE ADMISSION TYPES:")
print(df["Admission Type"].unique())

print("\nUNIQUE INSURANCE PROVIDERS:")
print(df["Insurance Provider"].unique())


# ============================================================
# 10. Admission Type Analysis
# ============================================================

print("\nADMISSION TYPES:")
print(df["Admission Type"].unique())

print("\nADMISSION TYPE FREQUENCY:")
print(df["Admission Type"].value_counts())

# Categorize admissions by urgency
df["Urgency Category"] = df["Admission Type"].map({
    "Emergency": "High Urgency",
    "Urgent": "Medium Urgency",
    "Elective": "Low Urgency"
})

print("\nURGENCY CATEGORY FREQUENCY:")
print(df["Urgency Category"].value_counts())


# ============================================================
# 11. Length of Stay Analysis
# ============================================================

df["Length of Stay"] = (
    df["Discharge Date"] -
    df["Date of Admission"]
).dt.days

print("\nLENGTH OF STAY STATISTICS:")
print(df["Length of Stay"].describe())

print("\nMINIMUM STAY:")
print(df["Length of Stay"].min())

print("\nMAXIMUM STAY:")
print(df["Length of Stay"].max())

print("\nAVERAGE STAY:")
print(df["Length of Stay"].mean())

print("\nMEDIAN STAY:")
print(df["Length of Stay"].median())

print("\nSTANDARD DEVIATION OF STAY:")
print(df["Length of Stay"].std())


# ============================================================
# 12. Billing Analysis
# ============================================================

print("\nTOTAL BILLING:")
print(df["Billing Amount"].sum())

print("\nAVERAGE BILLING:")
print(df["Billing Amount"].mean())

print("\nMEDIAN BILLING:")
print(df["Billing Amount"].median())

print("\nMINIMUM BILLING:")
print(df["Billing Amount"].min())

print("\nMAXIMUM BILLING:")
print(df["Billing Amount"].max())

print("\nBILLING STANDARD DEVIATION:")
print(df["Billing Amount"].std())


# ============================================================
# 13. Demographic Analysis
# ============================================================

print("\nAGE STATISTICS:")
print(df["Age"].describe())

print("\nGENDER-WISE PATIENT COUNT:")
print(df["Gender"].value_counts())

# Create age groups
df["Age Group"] = pd.cut(
    df["Age"],
    bins=[0, 18, 30, 45, 60, 75, 100],
    labels=[
        "0-18",
        "19-30",
        "31-45",
        "46-60",
        "61-75",
        "76-100"
    ]
)

print("\nAGE-GROUP DISTRIBUTION:")
print(df["Age Group"].value_counts().sort_index())


# ============================================================
# 14. Medical Condition Analysis
# ============================================================

print("\nPATIENT COUNT BY MEDICAL CONDITION:")
print(df["Medical Condition"].value_counts())

print("\nAVERAGE AGE BY MEDICAL CONDITION:")
display(
    df.groupby("Medical Condition")["Age"]
    .mean()
    .reset_index()
)

print("\nADMISSION PATTERN BY MEDICAL CONDITION:")
display(
    pd.crosstab(
        df["Medical Condition"],
        df["Admission Type"]
    )
)

print("\nMEDICAL CONDITION SUMMARY:")

condition_summary = (
    df.groupby("Medical Condition")
    .agg(
        Patient_Count=("Name", "count"),
        Average_Age=("Age", "mean"),
        Median_Age=("Age", "median"),
        Average_Billing=("Billing Amount", "mean"),
        Average_Stay=("Length of Stay", "mean")
    )
    .reset_index()
)

display(condition_summary)


# ============================================================
# 15. Final Data Quality Check
# ============================================================

print("\nFINAL DATASET SHAPE:")
print(df.shape)

print("\nFINAL DATA TYPES:")
print(df.dtypes)

print("\nFINAL MISSING VALUE CHECK:")
display(df.isnull().sum())

print("\nFINAL DUPLICATE CHECK:")
print(df.duplicated().sum())

print("\nFINAL CATEGORICAL VALUES:")

print("\nGender:")
print(df["Gender"].unique())

print("\nMedical Condition:")
print(df["Medical Condition"].unique())

print("\nAdmission Type:")
print(df["Admission Type"].unique())

print("\nInsurance Provider:")
print(df["Insurance Provider"].unique())

print("\nFINAL NUMERICAL SUMMARY:")
display(
    df[
        [
            "Age",
            "Billing Amount",
            "Room Number",
            "Length of Stay"
        ]
    ].describe()
)

print("\nFINAL DATASET:")
display(df.head())


# ============================================================
# 16. Save Preprocessed Dataset for Week 6
# ============================================================

df.to_csv(
    "healthcare_preprocessed.csv",
    index=False
)

print("\nPreprocessed dataset saved as:")
print("healthcare_preprocessed.csv")
```
