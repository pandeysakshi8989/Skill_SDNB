## 1. What is Data Analysis?

**Data Analysis** is the process of examining, cleaning, transforming and interpreting data to discover useful information, patterns and insights.

A Data Analyst generally works with data to answer questions such as:

* What happened?
* How much?
* Which category performed best?
* Which region generated more profit?
* Are there unusual values?
* What patterns can we observe?
* What conclusions can we draw from the data?

### Simple Example

Suppose a company has thousands of sales records.

The raw data may contain:

* Order details
* Customer information
* Product information
* Sales
* Quantity
* Discount
* Profit
* Region

A Data Analyst can analyse this data to answer:

> Which product category generates the highest sales?

> Which region generates the highest profit?

> What is the average order value?

> Are there products with high sales but low profit?

Therefore, Data Analysis helps convert **raw data into useful information and insights**.

**PART 1: DAY 4 — Theoretical Foundation**

1. The Split-Apply-Combine Paradigm
Data aggregation in Pandas relies on three sequential phases:

Split: Divides the DataFrame into smaller subsets based on unique values of specified key column(s).

Apply: Computes aggregate calculations (e.g., sum, mean, count, custom lambdas) independently on each group.

Combine: Aggregates individual output vectors back into a single tabular DataFrame or Series.

| Function / Method | Syntax Example | Primary Purpose |
| --- | --- | --- |
| `groupby()` | `df.groupby('Category')` | Splits dataset into lazy `DataFrameGroupBy` objects. |
| `agg()` | `df.groupby('Region').agg(Total=('Sales', 'sum'))` | Executes flexible, custom multi-column aggregations. |
| `sort_values()` | `df.sort_values(by='Sales', ascending=False)` | Orders rows in ascending/descending direction. |
| `reset_index()` | `df.groupby('Region')['Sales'].sum().reset_index()` | Flattens grouped index back into regular columns. |
| `filter()` | `df.groupby('City').filter(lambda x: x['Sales'].sum() > 50000)` | Keeps or discards entire groups based on group-level conditions. |
| `nlargest()` | `df.nlargest(5, 'Sales')` | Retrieves top $N$ records optimized without full heap sorting. |
| `size()` vs `count()` | `df.groupby('Region').size()` | `size()` counts all rows (including `NaN`); `count()` ignores missing values. |

## IMPORTING LIBRARIES AND  LOADING DATASETS

````python
# SUPERSTORE DATASET

import kagglehub

# Download latest version
path = kagglehub.dataset_download("himanshuuike/superstore-sales-dataset")

print("Path to dataset files:", path)

import os

print(os.listdir(path))

import pandas as pd

df_superstore = pd.read_csv(os.path.join(path, "samplesuperstore.csv"))

print( "Dataset loaded successfully")
````

**Task 1: Single-Column GroupBy Aggregation**

Question: Calculate the total sales generated by each Region.

````python
# Code
task1_superstore = df_superstore.groupby('Region')['Sales'].sum().reset_index()
print(task1_superstore)
````

Observation: The West region generates the highest revenue (~$725,457), while the South region accounts for the lowest sales volume (~$391,721).

**Task 2: Multi-Column GroupBy Aggregation**

Question: Calculate total Sales and Profit broken down by Category and Sub-Category.

````python
# Code
task2_superstore = df_superstore.groupby(['Category', 'Sub-Category'])[['Sales', 'Profit']].sum().reset_index()
print(task2_superstore.head(5))
````

Observation: Despite generating high sales (> $200,000), Sub-Categories like Tables and Bookcases operate at significant cumulative net losses.

**Task 3: Multi-Metric Aggregation with agg()**

Question: Compute average Sales, minimum Profit, and maximum Profit for each Product Category.

````python
# Code
task3_superstore = df_superstore.groupby('Category').agg(
    Avg_Sales=('Sales', 'mean'),
    Min_Profit=('Profit', 'min'),
    Max_Profit=('Profit', 'max')
).reset_index()
print(task3_superstore)
````

Observation: Technology has both the highest single order profit ($8,399.97) and the average transaction size ($452.71), alongside extreme downside variance.

**Task 4: Sorting Aggregated Summaries**

Question: Identify the Top 5 Sub-Categories generating the highest cumulative Profit.


````python
# Code
task4_superstore = (df_superstore.groupby('Sub-Category')['Profit']
                    .sum()
                    .reset_index()
                    .sort_values(by='Profit', ascending=False)
                    .head(5))
print(task4_superstore)
````

Observation: Copiers represent the most lucrative sub-category, yielding over $55,617 in profit despite lower unit order frequency.

**Task 5: Named Aggregations**

Question: Calculate total Order Count, average Discount rate, and total Profit across Customer Segments.

````python
# Code
task5_superstore = df_superstore.groupby('Segment').agg(
    Total_Orders=('Row ID', 'count'),
    Avg_Discount=('Discount', 'mean'),
    Total_Profit=('Profit', 'sum')
).reset_index()
print(task5_superstore)
````

Observation: Consumers make up more than 50% of overall order volume, maintaining consistent average discount ratios across all segments (~15%).

**Task 6: Conditional Group Filtering**

Question: Filter the dataset to isolate rows originating only from Cities whose total Sales exceed $50,000.

````python
# Code
task6_superstore = df_superstore.groupby('City').filter(lambda x: x['Sales'].sum() > 50000)
print("Qualified Cities:", task6_superstore['City'].unique())
print("Retained Rows Ratio:", f"{len(task6_superstore)} / {len(df_superstore)}")
````

Observation: Only 5 major metropolitan areas cross the $50,000 threshold, accounting for over 20% of all transaction volume.

**Task 7: Derived Aggregate Calculations**

Question: Calculate total Sales, total Profit, and Profit Margin percentage (Profit / Sales * 100) per Ship Mode.

````python
# Code
task7_superstore = df_superstore.groupby('Ship Mode').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum')
).reset_index()
task7_superstore['Profit_Margin_%'] = (task7_superstore['Total_Profit'] / task7_superstore['Total_Sales']) * 100
print(task7_superstore.sort_values(by='Profit_Margin_%', ascending=False))
````

Observation: First Class shipping yields the highest profit margin (13.93%), whereas Standard Class carries the lowest margin despite dominating volume.

**Task 8: Extremum Selection with nlargest()**

Question: Extract the top 5 individual transaction orders based on gross Sales value.

````python
# Code
task8_superstore = df_superstore.nlargest(5, 'Sales')[['Order ID', 'Category', 'Sub-Category', 'Sales', 'Profit']]
print(task8_superstore)
````

Observation: The top 5 largest transactions belong exclusively to Technology and Office Supplies (Binders), with individual sales reaching up to $22,638.

**Task 9: Quantile Aggregation**

Question: Find the 75th percentile (Q3) of order Sales across each Product Category.


````python
# Code
task9_superstore = df_superstore.groupby('Category')['Sales'].agg(
    Q3_Sales=lambda x: x.quantile(0.75)
).reset_index()
print(task9_superstore)
````

Observation: 75% of all Office Supplies transactions are priced below $51.98, while Furniture and Technology have much higher 75th percentile cuts (~$335).

**Task 10: Frequency Counting via GroupBy**

Question: Count total order line items across combinations of Region and Customer Segment.

````python
# Code
task10_superstore = (df_superstore.groupby(['Region', 'Segment'])
                     .size()
                     .reset_index(name='Order_Count')
                     .sort_values(by='Order_Count', ascending=False))
print(task10_superstore.head(5))
````

Observation: The West Consumer combination represents the single highest volume operational segment (1,672 order items).

#### Task 2: Calculate and Plot 7-Day Moving Average of Sales
**AI Prompt for Code Generation:**
"Calculate a 7-day rolling mean of daily Sales and plot original daily sales alongside rolling average."

````python
import matplotlib.pyplot as plt

# Ensure 'Order Date' is datetime and set as index. This was done in the previous cell
# df['Order Date'] = pd.to_datetime(df['Order Date'])
# df_superstore_time = df.set_index('Order Date')

# Aggregate daily sales (sum, in case there are multiple orders per day)
daily_sales = df_superstore_time['Sales'].resample('D').sum()

# Calculate the 7-day rolling mean of daily sales
rolling_mean_7_day = daily_sales.rolling(window=7).mean()

# Plot original daily sales and the 7-day rolling mean
plt.figure(figsize=(14, 7))
plt.plot(daily_sales, label='Daily Sales', alpha=0.7)
plt.plot(rolling_mean_7_day, label='7-Day Rolling Mean', color='red')
plt.title('Daily Sales and 7-Day Rolling Mean (Superstore)')
plt.xlabel('Order Date')
plt.ylabel('Total Sales')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
````

#### Task 2: Matplotlib 3D Surface Plot of Synthetic Sales-Profit Surface
**AI Prompt for Code Generation:**
"Generate a smooth 3D Surface plot using np.meshgrid and ax.plot_surface in Matplotlib."

````python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Define the range for X and Y values
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

# Create a meshgrid
X, Y = np.meshgrid(x, y)

# Define a function for Z values (e.g., a 3D sine wave)
Z = np.sin(np.sqrt(X**2 + Y**2))

# Create a figure and a 3D axes object
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
surface = ax.plot_surface(X, Y, Z, cmap='viridis', rstride=1, cstride=1, alpha=0.8)

# Set labels
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')

# Set title
ax.set_title('Smooth 3D Surface Plot')

# Add a color bar
fig.colorbar(surface, shrink=0.5, aspect=10)

plt.tight_layout()
plt.show()
````

**Observation:**
3D Surface charts render dynamic continuous topographies that are ideal for modeling response surfaces and profit projections.

"Build a single integrated Plotly Dash application with a Region dropdown filter that dynamically updates dynamic KPI cards, a Sales by Category bar chart, a Profit vs Sales scatter plot, a Discount histogram, and a Category Sales pie chart."

````python
# ============================================================
# PART 3: INTERACTIVE DASHBOARD SUITE USING PLOTLY DASH
# TASKS 6 TO 10
# ============================================================

import dash
from dash import dcc, html, Input, Output
import plotly.express as px
import pandas as pd


# ============================================================
# STEP 1: PREPARE df_superstore
# ============================================================

df_superstore["Sales"] = pd.to_numeric(
    df_superstore["Sales"],
    errors="coerce"
)

df_superstore["Profit"] = pd.to_numeric(
    df_superstore["Profit"],
    errors="coerce"
)

df_superstore["Discount"] = pd.to_numeric(
    df_superstore["Discount"],
    errors="coerce"
)

df_superstore = df_superstore.dropna(
    subset=["Region", "Category", "Sales", "Profit", "Discount"]
)


# ============================================================
# STEP 2: CREATE DASH APP
# ============================================================

app = dash.Dash(__name__)


# ============================================================
# STEP 3: DASHBOARD LAYOUT
# ============================================================

app.layout = html.Div(

    [

        # ----------------------------------------------------
        # Dashboard Title
        # ----------------------------------------------------

        html.H1(
            "Superstore Sales Interactive Dashboard",
            style={
                "textAlign": "center"
            }
        ),

        html.P(
            "Interactive analysis using df_superstore",
            style={
                "textAlign": "center"
            }
        ),


        # ----------------------------------------------------
        # TASK 6: REGION DROPDOWN
        # ----------------------------------------------------

        html.Div(

            [

                html.Label(
                    "Select Region:",
                    style={
                        "fontWeight": "bold",
                        "fontSize": "18px"
                    }
                ),

                dcc.Dropdown(

                    id="region-dropdown",

                    options=[
                        {
                            "label": region,
                            "value": region
                        }

                        for region
                        in sorted(
                            df_superstore["Region"].unique()
                        )
                    ],

                    value="Central",

                    clearable=False

                )

            ],

            style={
                "width": "50%",
                "margin": "auto",
                "marginBottom": "30px"
            }
        ),


        # ----------------------------------------------------
        # TASK 10: KPI CARDS
        # ----------------------------------------------------

        html.Div(

            [

                # Total Sales
                html.Div(
                    [
                        html.H3("Total Sales"),

                        html.H2(
                            id="total-sales"
                        )
                    ],

                    style={
                        "textAlign": "center",
                        "padding": "20px",
                        "backgroundColor": "white",
                        "borderRadius": "10px",
                        "width": "30%"
                    }
                ),


                # Total Profit
                html.Div(
                    [
                        html.H3("Total Profit"),

                        html.H2(
                            id="total-profit"
                        )
                    ],

                    style={
                        "textAlign": "center",
                        "padding": "20px",
                        "backgroundColor": "white",
                        "borderRadius": "10px",
                        "width": "30%"
                    }
                ),


                # Number of Records
                html.Div(
                    [
                        html.H3("Number of Records"),

                        html.H2(
                            id="total-records"
                        )
                    ],

                    style={
                        "textAlign": "center",
                        "padding": "20px",
                        "backgroundColor": "white",
                        "borderRadius": "10px",
                        "width": "30%"
                    }
                )

            ],

            style={
                "display": "flex",
                "justifyContent": "space-around",
                "marginBottom": "30px"
            }
        ),


        # ----------------------------------------------------
        # TASK 7: SALES BY CATEGORY BAR CHART
        # ----------------------------------------------------

        html.Div(

            [

                dcc.Graph(
                    id="sales-category-bar"
                )

            ],

            style={
                "marginBottom": "30px"
            }
        ),


        # ----------------------------------------------------
        # TASK 8: PROFIT VS SALES SCATTER
        # ----------------------------------------------------

        html.Div(

            [

                dcc.Graph(
                    id="profit-sales-scatter"
                )

            ],

            style={
                "marginBottom": "30px"
            }
        ),


        # ----------------------------------------------------
        # TASK 9: DISCOUNT HISTOGRAM
        # ----------------------------------------------------

        html.Div(

            [

                dcc.Graph(
                    id="discount-histogram"
                )

            ],

            style={
                "marginBottom": "30px"
            }
        ),


        # ----------------------------------------------------
        # TASK 10: CATEGORY SALES PIE CHART
        # ----------------------------------------------------

        html.Div(

            [

                dcc.Graph(
                    id="category-sales-pie"
                )

            ]

        )

    ],

    style={
        "padding": "30px",
        "backgroundColor": "#f5f5f5"
    }
)


# ============================================================
# STEP 4: CALLBACK
# ============================================================

@app.callback(

    [
        Output("total-sales", "children"),
        Output("total-profit", "children"),
        Output("total-records", "children"),
        Output("sales-category-bar", "figure"),
        Output("profit-sales-scatter", "figure"),
        Output("discount-histogram", "figure"),
        Output("category-sales-pie", "figure")
    ],

    Input(
        "region-dropdown",
        "value"
    )
)


def update_dashboard(selected_region):

    # ========================================================
    # FILTER df_superstore
    # ========================================================

    filtered_df = df_superstore[
        df_superstore["Region"] == selected_region
    ]


    # ========================================================
    # KPI CALCULATIONS
    # ========================================================

    total_sales = filtered_df["Sales"].sum()

    total_profit = filtered_df["Profit"].sum()

    total_records = len(filtered_df)


    # ========================================================
    # TASK 7: SALES BY CATEGORY BAR CHART
    # ========================================================

    category_sales = (

        filtered_df
        .groupby("Category")["Sales"]
        .sum()
        .reset_index()

    )


    bar_fig = px.bar(

        category_sales,

        x="Category",

        y="Sales",

        title=f"Sales by Category - {selected_region}",

        text_auto=True

    )


    bar_fig.update_layout(

        xaxis_title="Category",

        yaxis_title="Total Sales"

    )


    # ========================================================
    # TASK 8: PROFIT VS SALES SCATTER PLOT
    # ========================================================

    scatter_fig = px.scatter(

        filtered_df,

        x="Sales",

        y="Profit",

        color="Category",

        title=f"Profit vs Sales - {selected_region}",

        hover_data=[
            "Category",
            "Discount"
        ]

    )


    scatter_fig.add_hline(
        y=0,
        line_dash="dash"
    )


    # ========================================================
    # TASK 9: DISCOUNT HISTOGRAM
    # ========================================================

    histogram_fig = px.histogram(

        filtered_df,

        x="Discount",

        title=f"Discount Distribution - {selected_region}",

        nbins=10

    )


    histogram_fig.update_layout(

        xaxis_title="Discount",

        yaxis_title="Number of Records"

    )


    # ========================================================
    # TASK 10: CATEGORY SALES PIE CHART
    # ========================================================

    pie_fig = px.pie(

        category_sales,

        names="Category",

        values="Sales",

        title=f"Sales Distribution by Category - {selected_region}",

        hole=0.3

    )


    # ========================================================
    # RETURN ALL DASHBOARD COMPONENTS
    # ========================================================

    return (

        f"${total_sales:,.2f}",

        f"${total_profit:,.2f}",

        f"{total_records:,}",

        bar_fig,

        scatter_fig,

        histogram_fig,

        pie_fig

    )


# ============================================================
# STEP 5: RUN APPLICATION
# ============================================================

app.run(debug=True)
````
